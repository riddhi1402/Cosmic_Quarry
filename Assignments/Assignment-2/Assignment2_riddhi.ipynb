```python
import numpy as np


arr = np.array([5, 8, 12, 15, 18, 22, 25, 30, 35, 40])

mean = np.mean(arr)
std_dev = np.std(arr)

print("Array:", arr)
print("Mean:", mean)
print("Standard Deviation:", std_dev)
```

```python
import pandas as pd
df = pd.read_csv('data.csv')
print(df)



```

```python
import matplotlib.pyplot as plt


plt.hist(df['Marks'], bins=10, color='purple', edgecolor='black')


plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Distribution of Marks')


plt.show()
```

```python

filtered_df = df[df['Marks'] > 90]

# the filtered DataFrame
print(filtered_df)
```

```python
import matplotlib.pyplot as plt


plt.scatter(df['Age'], df['Marks'], color='blue', alpha=0.7)

plt.xlabel('Age')
plt.ylabel('Marks')
plt.title('Scatter Plot of Age vs Marks')

plt.show()
```
